```python
import numpy as np

name = np.array(["Rishabh Yadav", "Vansh", "Juber", "Harsh"])
marks = np.array([96, 56, 85, 69])

result = np.column_stack((name, marks))

c = marks.max()
i = marks.argmax() #argmax() geven maximum value 

print("Highest Marks:", c)
print("Topper Name:", name[i])
print(result)
```

```python
import numpy as np

name = np.array(["Rishabh Yadav", "Vansh", "Juber", "Harsh"])
marks = np.array([96, 56, 85, 69])

result = np.column_stack((name, marks))

c = marks.mean()
i = marks.argmax() #argmax() geven maximum value 

print("Highest Marks:", c)
print("Topper Name:", name[i])
print(result)
```
